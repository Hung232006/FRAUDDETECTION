```python
import matplotlib.pyplot as plt
```

```python
df["Class"].value_counts().plot(kind="bar")

plt.title("Distribution of Transactions")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))

df["Class"].value_counts().plot(kind="bar")

plt.title("Normal vs Fraud Transactions")
plt.xlabel("Class")
plt.ylabel("Count")

plt.show()
```

```python
fraud = df["Class"].value_counts()[1]
normal = df["Class"].value_counts()[0]

print("Normal:", normal)
print("Fraud:", fraud)
print("Fraud Percentage:", fraud / len(df) * 100)
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Amount"], bins=50)

plt.title("Transaction Amount Distribution")
plt.xlabel("Amount")
plt.ylabel("Frequency")

plt.show()
```

```python
print("Min:", df["Amount"].min())
print("Max:", df["Amount"].max())
print("Mean:", df["Amount"].mean())
print("Median:", df["Amount"].median())
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Time"], bins=50)

plt.title("Transaction Time Distribution")
plt.xlabel("Time")
plt.ylabel("Frequency")

plt.show()
```

```python
fraud = df[df["Class"] == 1]
normal = df[df["Class"] == 0]

plt.figure(figsize=(8,5))

plt.hist(normal["Amount"], bins=50, alpha=0.5, label="Normal")
plt.hist(fraud["Amount"], bins=50, alpha=0.5, label="Fraud")

plt.legend()

plt.xlabel("Amount")
plt.ylabel("Frequency")
plt.title("Amount Comparison")

plt.show()
```
